# 📚 Machine Learning & Data Science Portfolio Summary

This portfolio documents a comprehensive study of fundamental and advanced Machine Learning and Data Science techniques applied across diverse problem domains, including astrophysics, materials science, and medical diagnosis.

---

## 💻 Core Machine Learning & Statistical Modeling

| Category | Key Algorithms & Models | Concepts Demonstrated |
| :--- | :--- | :--- |
| **Supervised Learning** | **Regression:** Linear, Ridge, Lasso, Random Forest, SVR, Gradient Boosting. **Classification:** k-NN, SVM, Logistic Regression, Random Forest. | Model selection, solving high-stakes binary classification, weighted non-linear least squares fitting (Curve Fitting), and model hypothesis testing. |
| **Unsupervised Learning** | **Clustering:** K-Means, DBSCAN. **Anomaly Detection:** One-Class SVM, Isolation Forest. | Identifying optimal number of clusters ($k$), handling arbitrary cluster shapes, and detecting outliers/data anomalies. |
| **Programming/OOP** | **Class-based Design** | Creating modular, reusable code architectures for vector operations and ML workflows (**Encapsulation** and **Modularity**). |

---

## ⚙️ Optimization & Robust Workflow Design

* **Hyperparameter Tuning:** Implemented **Grid Search CV** (Cross-Validation) as the primary method for robust model optimization, ensuring the final model generalizes well.
* **ML Pipelines:** Used **`sklearn.pipeline`** to chain preprocessing (Standardization/Normalization) with modeling, effectively preventing **data leakage** during cross-validation.
* **Regularization:** Explored **L1 (Lasso)** and **L2 (Ridge)** regularization with Logistic Regression, demonstrating how L1 achieves **sparsity (automatic feature selection)** while L2 stabilizes coefficients.
* **Advanced Optimization:** Utilized **Bayesian Optimization (BO)** with a Random Forest surrogate model to efficiently search complex compositional spaces (materials science) under defined scientific constraints.

---

## 📊 Feature & Performance Analysis

* **Evaluation Metrics:** Assessed model performance using $\text{RMSE}$, $\text{MAE}$, $\chi^2$ (for fitting), **Accuracy**, **Precision**, and **Recall** (with focus on minimizing False Negatives in medical tasks).
* **Feature Importance:** Employed both **Brute Force Forward Selection** (performance-driven MAE reduction) and **SHAP** (Shapley Additive exPlanations) for transparent model interpretation and feature ranking.
* **Uncertainty & Visualization:** Generated **Classification Reports**, **Confusion Matrices**, and **ROC Curves**. Used Monte Carlo sampling and the covariance matrix to visualize **model uncertainty** for statistical fits.
* **Validation:** Compared model performance across different feature subsets (e.g., using only features critical to the Hertzsprung-Russell diagram) and utilized **Davies-Bouldin Index** and **Calinski-Harabasz Index** for quantitative cluster validation.
